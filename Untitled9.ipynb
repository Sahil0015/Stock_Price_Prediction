```python
def plot_graph(size, data, title):
    plt.figure(figsize=size)

    # Handle both Series and DataFrame
    if isinstance(data, pd.Series):
        plt.plot(data, label=data.name)
        plt.ylabel(data.name)  # Use series name for y-axis label
    elif isinstance(data, pd.DataFrame):
        # Loop through each column and plot them
        for col in data.columns:
            plt.plot(data[col], label=col)
        plt.ylabel("Values")  # Default label for DataFrame columns

    plt.xlabel("Years")
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(close_price_data.values.reshape(-1, 1))

scaled_data
```

```python
len(scaled_data)
```

```python
short_term_x_data = []
short_term_y_data = []

medium_term_x_data = []
medium_term_y_data = []

long_term_x_data = []
long_term_y_data = []

for i in range(10, len(scaled_data)):
  short_term_x_data.append(scaled_data[i-10:i])
  short_term_y_data.append(scaled_data[i])

for i in range(40, len(scaled_data)):
  medium_term_x_data.append(scaled_data[i-40:i])
  medium_term_y_data.append(scaled_data[i])

for i in range(100, len(scaled_data)):
  long_term_x_data.append(scaled_data[i-100:i])
  long_term_y_data.append(scaled_data[i])
```

```python
short_term_x_data = np.array(short_term_x_data)
short_term_y_data = np.array(short_term_y_data)

medium_term_x_data = np.array(medium_term_x_data)
medium_term_y_data = np.array(medium_term_y_data)

long_term_x_data = np.array(long_term_x_data)
long_term_y_data = np.array(long_term_y_data)
```

```python
short_splitting_len = int(len(short_term_x_data) * 0.8)
medium_splitting_len = int(len(medium_term_x_data) * 0.75)
long_splitting_len = int(len(long_term_x_data) * 0.7)

short_x_train = short_term_x_data[:short_splitting_len]
short_x_test = short_term_x_data[short_splitting_len:]
short_y_train = short_term_y_data[:short_splitting_len]
short_y_test = short_term_y_data[short_splitting_len:]

medium_x_train = medium_term_x_data[:medium_splitting_len]
medium_x_test = medium_term_x_data[medium_splitting_len:]
medium_y_train = medium_term_y_data[:medium_splitting_len]
medium_y_test = medium_term_y_data[medium_splitting_len:]

long_x_train = long_term_x_data[:long_splitting_len]
long_x_test = long_term_x_data[long_splitting_len:]
long_y_train = long_term_y_data[:long_splitting_len]
long_y_test = long_term_y_data[long_splitting_len:]
```

```python
def build_short_term_lstm(input_shape):
    model = Sequential()
    # First LSTM layer: using 32 units
    model.add(LSTM(64, return_sequences=True, input_shape=input_shape))
    model.add(Dropout(0.2))  # dropout to reduce overfitting

    # Second LSTM layer: using 16 units
    model.add(LSTM(32))
    model.add(Dropout(0.2))

    model.add(Dense(4))

    # Output layer: single neuron for regression (price prediction)
    model.add(Dense(1))

    model.compile(optimizer='adam', loss='mse')
    return model


def build_medium_term_lstm(input_shape):
    model = Sequential()
    # First LSTM layer with 64 units
    model.add(LSTM(64, return_sequences=True, input_shape=input_shape))
    model.add(Dropout(0.2))  # dropout to reduce overfitting

    # Second LSTM layer with 32 units
    model.add(LSTM(32, return_sequences=True))
    model.add(Dropout(0.2))

    # Third LSTM layer to summarize the sequences
    model.add(LSTM(16))
    model.add(Dropout(0.2))

    model.add(Dense(8))

    # Output layer
    model.add(Dense(1))

    model.compile(optimizer='adam', loss='mse')
    return model


def build_long_term_lstm(input_shape):
    model = Sequential()
    # First LSTM layer with a high number of units to capture long-term patterns
    model.add(LSTM(128, return_sequences=True, input_shape=input_shape))
    model.add(Dropout(0.3))

    # Second LSTM layer
    model.add(LSTM(64, return_sequences=True))
    model.add(Dropout(0.3))

    # Third LSTM layer to continue processing the sequences
    model.add(LSTM(32, return_sequences=True))
    model.add(Dropout(0.3))

    # Fourth LSTM layer to reduce the sequence into a final representation
    model.add(LSTM(16))
    model.add(Dropout(0.2))

    model.add(Dense(4))

    # Output layer
    model.add(Dense(1))

    model.compile(optimizer='adam', loss='mse')
    return model
```

```python
short_x_train.shape[1:]
```

```python
short_term_model = build_short_term_lstm(short_x_train.shape[1:])
short_term_model.summary()
```

```python
medium_term_model = build_medium_term_lstm(medium_x_train.shape[1:])
medium_term_model.summary()
```

```python
long_term_model = build_long_term_lstm(long_x_train.shape[1:])
long_term_model.summary()
```

```python
checkpoint = ModelCheckpoint('model.h5', monitor='val_loss', save_best_only=True, mode='min')
early_stop = EarlyStopping(monitor='val_loss', patience=2, restore_best_weights=True)
```

```python
short_term_model.fit(short_x_train, short_y_train, epochs=25, batch_size=32,
          validation_data=(short_x_test, short_y_test), callbacks=[checkpoint, early_stop])
```

```python
medium_term_model.fit(medium_x_train, medium_y_train, epochs=25, batch_size=32,
          validation_data=(medium_x_test, medium_y_test), callbacks=[checkpoint, early_stop])
```

```python
long_term_model.fit(long_x_train, long_y_train, epochs=25, batch_size=32,
          validation_data=(long_x_test, long_y_test), callbacks=[checkpoint, early_stop])
```

```python
short_term_mae = mean_absolute_error(inv_short_y_test, inv_short_term_predictions)

short_term_mse = mean_squared_error(inv_short_y_test, inv_short_term_predictions)
short_term_rmse = np.sqrt(short_term_mse)

short_term_r2 = r2_score(inv_short_y_test, inv_short_term_predictions)

# Print the results
print("Short term trading Mean Absolute Error (MAE):", short_term_mae)
print("Short term trading Root Mean Squared Error (RMSE):", short_term_rmse)
print("Short term trading R² Score:", short_term_r2)
```

```python
medium_term_mae = mean_absolute_error(inv_medium_y_test, inv_medium_term_predictions)

medium_term_mse = mean_squared_error(inv_medium_y_test, inv_medium_term_predictions)
medium_term_rmse = np.sqrt(medium_term_mse)

medium_term_r2 = r2_score(inv_medium_y_test, inv_medium_term_predictions)

# Print the results
print("Medium term trading Mean Absolute Error (MAE):", medium_term_mae)
print("Medium term trading Root Mean Squared Error (RMSE):", medium_term_rmse)
print("Medium term trading R² Score:", medium_term_r2)
```

```python
long_term_mae = mean_absolute_error(inv_long_y_test, inv_long_term_predictions)

long_term_mse = mean_squared_error(inv_long_y_test, inv_long_term_predictions)
long_term_rmse = np.sqrt(long_term_mse)

long_term_r2 = r2_score(inv_medium_y_test, inv_medium_term_predictions)

# Print the results
print("Long term trading Mean Absolute Error (MAE):", long_term_mae)
print("Long term trading Root Mean Squared Error (RMSE):", long_term_rmse)
print("Long term trading R² Score:", long_term_r2)
```
